Now we are going to summarise different metrics results on both test and train data sets.

```python
ET = joblib.load('../artefacts/ET.joblib')
RF = joblib.load('../artefacts/RF.joblib')
GB = joblib.load('../artefacts/GB.joblib')
KNN = joblib.load('../artefacts/KNN.joblib')

X_test, y_test = read_test_data()
X_train, y_train = read_train_data()
```

```python
from sklearn.metrics import confusion_matrix, recall_score, accuracy_score, balanced_accuracy_score, roc_auc_score
```

a) ExtraTrees

```python
def model_evaluation_summary(model, model_name):

    model_train_pr = model.predict(X_train)
    model_test_pr = model.predict(X_test)

    model_train_proba = model.predict_proba(X_train)[:,1]
    model_test_proba = model.predict_proba(X_test)[:,1]

    print(f'---{model_name}---')
    print('ACCURACY')
    print(f'Train: {accuracy_score(y_pred=model_train_pr, y_true=y_train)}')
    print(f'Test: {accuracy_score(y_pred=model_test_pr, y_true=y_test)}\n')
    print('BALANCED Accuracy')
    print(f'Train: {balanced_accuracy_score(y_pred=model_train_pr, y_true=y_train)}')
    print(f'Test: {balanced_accuracy_score(y_pred=model_test_pr, y_true=y_test)}\n')
    print('RECALL')
    print(f'Train: {recall_score(y_pred=model_train_pr, y_true=y_train)}')
    print(f'Test: {recall_score(y_pred=model_test_pr, y_true=y_test)}\n')
    print('CONFUSION MATRIX')
    print(f'Train: {confusion_matrix(y_pred=model_train_pr, y_true=y_train)}')
    print(f'Test: {confusion_matrix(y_pred=model_test_pr, y_true=y_test)}\n')
    print('AUC')
    print(f'Train: {roc_auc_score(y_score=model_train_proba, y_true=y_train)}')
    print(f'Test: {roc_auc_score(y_score=model_test_proba, y_true=y_test)}\n')
```

```python
model_evaluation_summary(ET, 'ET')
```

```python
model_evaluation_summary(RF, 'RF')
```

```python
model_evaluation_summary(KNN, 'KNN')
```

```python
model_evaluation_summary(GB, 'GB')
```

Chosen model turned out to be the best one. However, KNN also performs well. It's practically a draw.
